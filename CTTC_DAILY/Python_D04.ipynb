## range() : it is a built in function in python that is used to print the data in given interval

```python
for i in range (1,101):
    print(i,end=" ")
```

```python
for i in range(1,101,2):
    print(i,end=" ")
```

```python
# range(start,end+1,step)
for i in range(100,0,-1):
    print(i,end=' ')
```

```python
for i in range(198,102,-3):
    print(i, end=" ")
```

```python
n=int(input("Enter the multiplication number"))
for i in range (1,11,1):
      print(f"{i} x {n} = {i*n}")
```

```python
for i in range(200,100,-1):
    if(i%3==0):
        print(i, end=' ')
```

```python
n=int(input("Enter the multiplication number"))
for i in range (2,int(n**0.5)+1):
    if n%i==0:
        print("composite")
        break
   
else:
       print ('prime')
      
```

```python
row=int(input("enter"))
for i in range(0,row+1):
    print(" "*(row-i),"* "*i)
```

```python
row=int(input("enter"))
for i in range(0,row):
    print(" "*(row-i),"* "*i)
for i in range(row,0,-1):
    print(" "*(row-i),"* "*i)
```

## Python while loop
<pre>
    python while loop is used to run a block of code untill a certain codition is met

    syntax of while loop is :
    initialisation 
    while condition:
        #body of while
        increment/decrement
</pre>

```python
i=200
while i>99:
    if i%5==0:
        print(i,end=' ')
    i-=1    
   
```
